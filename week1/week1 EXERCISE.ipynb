```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
MODEL_GEMINI = 'gemini-2.5-flash'
BASE_GEMINI_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
# set up environment
load_dotenv(override=True)
gemini_api_key = os.getenv('GOOGLE_API_KEY')

if not gemini_api_key:
    raise ValueError("GOOGLE_API_KEY is not set")
elif len(gemini_api_key) < 32:
    raise ValueError("GOOGLE_API_KEY invalid")

openai = OpenAI(base_url=BASE_GEMINI_URL, api_key=gemini_api_key)

# here is the question; type over this to ask something new
```

```python
# here is the question; type over this to ask something new

question = """
What is the best approach for identifying and integrating a reliable data source (such as weather data) from a website?
Please answer in a way that is easy to understand.
Respond in markdown without code blocks.
"""

system_prompt = """
You are a helpful assistant that can answer questions about the world. Make sure to treat me as a rookie. 
Don't mind if you use some fancy words, I'm sure I'll learn them later.
"""

user_prompt = f"""
Please figure out the answer to the following question:
{question}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming


stream = openai.chat.completions.create(model=MODEL_GEMINI, messages=[
    {"role": "system", "content": system_prompt}, 
    {"role": "user", "content": question}], 
    stream=True )

response = ""
# display the brochure in a typewriter animation
display_handle = display(Markdown(""), display_id=True)
# iterate over the stream and display the brochure in a typewriter animation
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)


```
